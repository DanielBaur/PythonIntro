## Strings

Strings are **sequences of characters**, for example:

````python
s = "Hello World"
````

Strings are represented by the type `str`. To create one, you can use single (`'`), double (`"`), or triple (`'''` or `"""`) quotation marks (the latter allow multiline strings), so the other kinds of quotation marks can appear inside the string:

### Strings Are Sequences

Because strings are sequences of elements (in this case, text characters), many operations that work with **sequences** can also be used with strings. These include:

````python
len(s) # Return the number of elements in the sequence
````

````python
s[0] # Return the element at index 0
````

````python
s + "!" # Sequences can be combined
````

## Lists and Tuples

While strings contain sequences of individual characters, **lists** and **tuples** can contain sequences of _arbitrary_ values. List elements can be changed, while tuples are immutable. Create a list by separating its elements with commas inside square brackets `[` and `]`, and a tuple by using parentheses `(` and `)`:

````python
l = [ 4, 0.5, "Alice" ]
l
````

````python
t = ( "Bob", True )
t
````

### Lists and Tuples Are Sequences

As with any sequence, we can determine the number of elements and access individual elements by their index:

````python
len(l), len(t)
````

````python
l[0] # Python indexing starts at index 0
````

````python
l[1]
````

````python
l[2]
````

````python
t[0]
````

````python
t[1]
````

### Lists Are Mutable

Unlike strings and tuples, lists can be modified by changing, adding, or removing elements:

````python
l[1] = -2.2 # Assign a new value to the list at index 1
l
````

````python
l.append(-3) # Add a new value to the end of the list
l
````

````python
l.pop() # Remove and return the last list element (or the element at the specified index)
````

## Slicing

You can access parts of a sequence using _slicing_ syntax:

```python
sequence[start:stop:step]
```

`start` is the first index to include, while `stop` is the first index not included in the slice. `step` specifies the interval between selected elements.

You do not need to provide all three arguments. By default, `start=0`, `stop=len(sequence)`, and `step=1`:

````python
l[:2] # By default, start is 0
````

````python
l[2:] # By default, stop is the end of the sequence
````

````python
l[::2] # Every second element
````

````python
l[::-1] # Reverse the sequence
````

> **Note:** Slicing is a powerful way to write concise code. You will encounter it frequently when working with _NumPy arrays_ in [Part IV - Numerical Computing with NumPy](Python%20Basics%20Part%20IV.ipynb).

## Dictionaries

Another important data type is the **dictionary**. A dictionary maps each _key_ to a _value_ and is created using curly braces `{` and `}`:

````python
d = { 'a':1, 'b':2, 'c':3 }
d
````

Access individual values by their key:

````python
d['a']
````

Dictionaries are mutable, like lists:

````python
d['d'] = 4
d
````

Dictionary keys do not have to be strings; keys and values can use different data types:

````python
e = { 'some_key': 4.2, 3517: 'some_value' }
e
````

````python
e['some_key']
````

````python
e[3517]
````
