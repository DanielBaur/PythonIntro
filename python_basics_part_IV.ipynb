Python lists are very flexible: they can contain values of different data types and can be easily modified (for example, with `append`). However, this flexibility comes at the cost of performance, so lists are not ideal for numerical calculations.

The **NumPy** module provides the n-dimensional array data type `numpy.ndarray`, which uses highly optimized C and Fortran code for numerical operations.

Arrays can contain values of only one numerical data type (for example, floating-point values) and are much less flexible than lists. For many scientific applications, such as working with datasets, this is exactly what we need.

By convention, we import NumPy using the alias `np`:

````python
import numpy as np # By convention, NumPy is imported with the alias `np`
````

## Creating NumPy Arrays

The easiest way to create NumPy arrays from Python lists is to use the `numpy.array` function:

````python
a = np.array([ 1, 2, 3, 5, 8, 13 ])
a
````

````python
b = np.array([ [ 1.5, 2.2, 3.1 ], [ 4.0, 5.2, 6.7 ] ])
b
````

NumPy arrays have several **attributes** that provide useful information about the array:

````python
a.ndim, b.ndim # The number of dimensions of each array
````

````python
a.shape, b.shape # The length of the array along each dimension
````

````python
a.dtype, b.dtype # The data type of each array
````

> **Reminder:** Use `<TAB>` completion and the `?` documentation in Jupyter Notebook whenever you are unsure which functions are available or what they do.

### There Are Many Ways to Create Arrays

- The `numpy.arange` function works similarly to Python's `range` function, but it can also take floating-point arguments:

````python
np.arange(10)
````

````python
np.arange(1.5, 2, 0.1)
````

- The `numpy.linspace` and `numpy.logspace` functions are also useful. They generate a specified number of values at linear or logarithmic intervals between two numbers:

````python
np.linspace(10, 20, 4)
````

````python
np.logspace(1, 3, 4)
````

- The `numpy.zeros` and `numpy.ones` functions create arrays filled with zeros or ones. To create multidimensional arrays, pass a tuple instead of an integer to their `shape` argument:

````python
np.zeros(5)
````

````python
np.ones((5, 2))
````

## Calculating with Arrays

Arrays can be combined **element-wise** using the standard operators `+`, `-`, `*`, `/`, and `**`:

````python
x = np.array([1,2,3])
y = np.array([4,5,6])
````

````python
x + 2 * y
````

````python
x ** y
````

> **Warning:** These operators are defined very differently for Python lists!

## Applying Functions to Arrays

Functions from the `math` module, such as `sin` and `exp`, operate on individual numbers. The functions with the same names in NumPy operate on arrays: **the function is applied to every array element** and is typically much faster than computing each element individually.

````python
phi = np.linspace(0, 2*np.pi, 10) # 10 values between 0 and 2π
np.sin(phi) # The sine of each of these values
````

NumPy also provides many functions that calculate properties of an array:

````python
x = np.linspace(0, 10, 100)
np.sum(x), np.mean(x), np.std(x)
````

These functions generalize to multiple dimensions by specifying the axis along which the calculation should be performed:

````python
x = np.array([ [ 1, 2 ], [ 3, 4 ] ])
np.sum(x), np.sum(x, axis=0), np.sum(x, axis=1)
````

## NumPy Arrays Are Sequences

We can use all functions defined for sequences on NumPy arrays:

````python
a = np.arange(3)
len(a)
````

````python
for x in a:
    print(x)
````

````python
a[0]
````

### Slicing Selects Parts of an Array

We have already seen the [slicing](Python%20Basics%20Part%20I.ipynb#slicing) syntax for sequences. It lets us access individual elements or parts of a sequence:

```python
a[start:stop:step]
```

NumPy extends this syntax to multidimensional arrays:

```python
b[start:stop:step, start:stop:step]
```

````python
x = np.arange(10)
````

````python
x[:5]
````

````python
x[::2]
````

Alternatively, instead of a single index, we can put a **list of indices** inside the subscript to select the corresponding array elements:

````python
x = np.array([ 1, 6, 4, 7, 9 ])
indices = [ 1, 0, 2, 1 ]
x[indices]
````

### Masking Filters an Array

NumPy also extends this syntax with **masking**. We provide an **array of Booleans** in the subscript, with the same length as the original array, and get only the elements whose mask value is `True`:

````python
x = np.array([ 1, 6, 4, 7, 9 ])
mask = np.array([ True, True, False, False, True ])
x[mask]
````

Masking is especially useful because **comparison operators** applied to NumPy arrays return Boolean arrays:

````python
x > 4
````

We can therefore filter an array to select the elements that satisfy a **condition**:

````python
x[x > 4]
````

Conditions can be combined using the `&` operator:

````python
x[(x > 4) & (x < 8)]
````

### Slices and Masks Can Also Be Assigned To

When a slice or mask of an array appears on the left-hand side of an assignment, the corresponding part of the original array is updated:

````python
x = np.array([ 1, 6, 4, 7, 9 ])
x[x > 4] = 0
x
````
