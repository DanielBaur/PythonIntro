## `if` Statements

The simplest form of _control flow_ is the `if` statement. A block of code runs only when a condition evaluates to the Boolean value `True`. An optional `else` block runs when the condition is _not_ `True`. The syntax for `if` statements is:

```python
if condition:
    # do something
elif condition:
    # do something else
else:
    # do yet something else
```

Notice that code blocks are not enclosed by special symbols. Instead, a condition ends with a colon (`:`), and its associated code block is **indented**.

> **Note:** In Python, code blocks are delimited by colons and indentation. By convention, each indentation level uses four spaces.

For example, change the value of `a` in the following code and see how the output changes.

````python
a = 1
if a < 5:
    print("a is too small; setting it to 5")
    a = 5
else:
    print("a is large enough.")
print("a is now {}.".format(a))
````

The first call to `print` and the change to the value of `a` happen only when `a` is less than `5`. Otherwise, the `else` block runs. The final call to `print` always runs.

### Comparison Operators Produce Booleans

The condition in an `if` statement can be any expression that returns a `bool` value.

Values of data types that support comparisons can be combined with comparison operators to produce a Boolean:

````python
1 == 3 # equal
````

````python
1 != 3 # not equal
````

````python
1 > 3 # greater than
````

````python
3 <= 3.4 # less than or equal to
````

> The simplest conditions are the explicit Boolean values `True` and `False`. Note that in Python, `True` and `False` are written with an uppercase first letter.

### Logical Operators Combine Booleans

Booleans can be combined using the logical operators `and`, `or`, and `not`:

````python
True and False
````

````python
True or False
````

````python
not True
````

````python
(False and (True or False)) or (False and True)
````

### Additional Boolean Operators

Python provides several other useful operators.

- The `in` operator checks whether a value is present in a sequence:

````python
3 in [ 1, 3, 5 ]
````

- The `is` operator checks whether two references point to the _same object_.

````python
s = "Hello World"
t = "Hello World"
s == t, s is t
````

````python
1000 == 10**3, 1000 is 10**3
````

> Most of the time, we are interested only in the _equality_ of values, so we use the equality operator (`==`). The _identity_ of objects becomes relevant in _object-oriented programming_ when working with _reference_ types rather than the _value_ types covered so far.
>
> A common use of the identity operator `is` is checking against `None`. In Python, `None` represents the absence of a value:

````python
a = None
if a is None:
    print("a has no value!")
else:
    print("a is {}.".format(a))
````

## `for`-Schleifen


A loop can execute a block of code multiple times. The most commonly used loop is the `for` loop, with this syntax:

```python
for value in iterable:
    # do things
```

`iterable` can be any _sequence_, such as a list, tuple, or string:

````python
for x in [3, 1.2, 'a']:
    print(x)
````

````python
for letter in 'hello':
    print(letter)
````

### Repeat a Block of Code `n` Times

Sometimes a block of code needs to run a fixed number of times. We can use the `range` function for this:

````python
for i in range(5):
    print(i)
````

You can inspect the definition of the `range` function and the options it provides by opening its `?` documentation. Remove the `#` character and run the cell:

````python
#range?
````

We can also call the function with the arguments `start`, `stop`, and (optionally) `step`:

````python
for i in range(2, 10, 2):
    print(i)
````

### Stop a Loop or Skip an Iteration

- The `break` statement stops a loop:

````python
for i in range(100):
    if i > 3:
        break
    print(i)
````

- The `continue` statement skips only the current iteration of a loop:

````python
for i in range(3):
    if i == 1:
        continue
    print(i)
````

### Create Lists with `for` Loops

A very useful way to work with lists in Python is called a _list comprehension_. The following syntax applies an operation to each element in a sequence and returns a list of the resulting elements:

```python
new_elements = [operation(element) for element in iterable]
```

For example, this applies the operation $x^2$ to every $x \in [0,10]$:

````python
[x**2 for x in range(11)]
````

## `while` Loops

A `while` loop repeatedly executes a block of code until a condition evaluates to `False`. Its syntax is:

```python
while condition:
    # do something
```

````python
a = 1
while a < 10:
    print("a is less than 10 ({}); multiplying by 1.5.".format(a))
    a = a * 1.5
print("Loop finished; a is now {}.".format(a))
````

#### 16th May 2025: Equivalent Class Verification

##### translating the SW requirements into a processable dictionary

````python
SW_requirements = {
    "req1" : {
        "cond" : lambda x: True,
        "add"  : 1500,
    },
    "req2" : {
        "cond" : lambda x: True,
        "add"  : 1,
    },
    "req3" : {
        "cond" : lambda x: True,
        "add"  : 1,
    },
    "req4" : {
        "cond" : lambda x: True,
        "add"  : 1,
    },
    "req5" : {
        "cond" : lambda x: True,
        "add"  : 1,
    },
    "req6" : {
        "cond" : lambda x: True,
        "add"   : 1,
    },
    "req7" : {
        "cond" : lambda x: True,
        "add"  : 1,
    },
    "req8" : {
        "cond" : lambda x: True,
        "add"  : 1,
    },
}
````

##### determining the elementary test cases

````python

A_values = ["A<25", "25<=A<45", "45<=A<=65", "65<A"]
G_values = ["W", "!W"]
C_values = ["Sp", "Fam", "Reg"]

import itertools

set_of_elementary_test_cases_T = list(itertools.product(A_values, G_values, C_values))
print(set_of_elementary_test_cases_T)
````

##### defining the SW to be tested

````python

def calc_insurnce_rate(
    client,          # a three-element iterable resembling a client according to above definition of elementary test cases, e.g., ('65<A', 'W', 'Sp')
    requirements,    # a dictionary resembling the to be tested SW, see above
):

    # initial rate equals zero
    rate = 0

    # checking which of the requirements applies to the client
    for req in [*requirements]:
        if requirements[req]["cond"](client) == True:
            rate += requirements[req]["add"]

    # returning the computed rate
    return rate


print(calc_insurnce_rate(client=('65<A', 'W', 'Sp'), requirements=SW_requirements))
````

##### computing the rates for all elementary test cases

````python

list_of_computed_rates = [[t[0], t[1], t[2], 0] for t in set_of_elementary_test_cases_T]

for i, t in enumerate(set_of_elementary_test_cases_T):
    list_of_computed_rates[i][3] = calc_insurnce_rate(client=t, requirements=SW_requirements)

print(list_of_computed_rates)
````

##### testing whether our equivalent classes are complete

````python
applied_rates = []
for i, t in enumerate(list_of_computed_rates):
    applied_rates.append(t[3])
applied_rates = list(set(applied_rates))
print(f"In total, {len(applied_rates)} different rates have been applied: {applied_rates}.")

````
