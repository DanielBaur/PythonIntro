# Python Basics Part V - Plotting with Plotly

Now that we can analyze data numerically with NumPy, we can use **Plotly** (https://plotly.com) to create publication-quality scientific plots and visualize our results.

We could use the Plotly Express framework to create plots without much configuration. To make plots more consistent, this course provides a `plotting_helper.py` script with reusable plotting functions. Take a look at its source code to understand how these functions work.

First, import the helper functions:

```python
# Import the plotting_helper functions
from plotting_helper import *

# NumPy is also needed.
import numpy as np
```

```python
X = np.arange(0, 10, 1)
Y = np.exp(X)
```

```python
# Use styled_figure and dot_scatter to create a scatter plot.
# Customize the title and axis labels when calling the function.
fig0 = styled_figure(
    data=[dot_scatter(x=X,y=Y, name='X-Y0')], 
    title='X vs. Y0',
    title_x='X',
    title_y='Y0',
)
fig0.show()
```

To add another plot to this figure, use the `add_trace` function.

```python
Y1 = np.sqrt(X)*1000
fig0.add_trace(dot_scatter(x=X,y=Y1, name='X-Y1'))
```
