However, a program can quickly become long and difficult to read when we try to perform more complex calculations. An important programming principle is to **avoid repetition**. To do so, we define reusable **functions** and use existing functionality from **modules**.

## Functions Are the Building Blocks of a Program

A function is a block of code that performs a well-defined task. Functions always have a **name**, can accept **arguments**, and can return **values**. In Python, functions use this syntax:

```python
def function_name(arguments):
    # code here
    return values
```

As with `if` statements and loops, indentation marks the function's code block.

Once defined, a function can be called using this syntax:

```python
function_name(arguments)
```

Functions let us break a complex problem into smaller, solvable parts, then combine those parts into a complete program.

> **Example:** Once we have written a function that sorts a list, we can reuse it instead of writing the same code again each time.

> **Note:** You _can_ put code into functions, but that does not mean you always should. Write a function when it avoids repetition or makes the program's structure clearer. **A function is often useful when you can give it a descriptive name.**

### Arguments and Return Values

A function can accept multiple arguments...

````python
def add(a, b):
    return a + b

print(add(1,3))
print(add(1.,3.2))
print(add(4,3.))
````

...and it can return multiple values:

````python
def double_and_halve(value):
    return value * 2, value / 2

print(double_and_halve(5))
````

Return values can be assigned to one or more variables:

````python
d, h = double_and_halve(5.)
print(d)
print(h)
````

Functions can **call other functions**:

````python
def do_a():
    print("doing A")
    
def do_b():
    print("doing B")
    
def do_a_and_b():
    do_a()
    do_b()
    
do_a_and_b()
````

Arguments can have a **default value**, which makes them **optional**:

````python
def say_hello(to_name="World"):
    print("Hello {}!".format(to_name))
say_hello()
say_hello("Alice")
````

Arguments can be passed in the **order** in which the function defines them, or in any order by specifying the argument names:

### One-Line `lambda` Functions Are Useful for Mathematics

With `lambda` syntax, we can define a function on a single line:

```python
function_name = lambda arguments: return_value
```

This is often useful for mathematical functions:

````python
linear = lambda x, a, b: a * x + b
linear(0, a=1, b=1)
````

### Variables Are Available Where They Are Defined

Function arguments and variables defined inside a function's code block are available only within that function (**_local scope_**):

````python
def do_something():
    local_var = 1 # This variable is defined inside the function...
# ...and is not available outside it:
print(local_var)
````

Variables defined outside a function before it is called are also available inside the function (**_global scope_**):

````python
PI = 3.14 # This variable is defined globally...
# ...and can be used inside functions:
degrees_to_radians = lambda degrees: degrees / 180 * PI
degrees_to_radians(90)
````

> **In functions, use global-scope variables only when they remain constant while the program runs.** If a function needs input values, pass them as arguments.
>
> By convention, constants in Python are written in uppercase, as in `PI`.

### Local Variables Take Precedence

If a variable is defined both globally and locally, the local variable takes precedence:

````python
a = 1
def show_var():
    a = 2
    print(a)
show_var() # The local variable is used here...
print(a) # ...and the global variable is used here.
````

## Importing Functions from Modules

The good news is that many problems have already been solved. For common tasks such as sorting lists, highly optimized and tested solutions are available, and we should use them instead of writing our own.

Apart from some basic data types and functions such as `print` and `len`, these functions are organized into **modules** rather than being built into Python. We can **import** a module to access its functions:

```python
import module
module.function_name()
```

We can give frequently used modules a shorter name:

```python
import module as m
m.function_name()
```

We can also import individual functions from a module:

```python
from module import function_name
function_name()
```

Instead of writing the `factorial` function from the previous exercise ourselves, we can use the function of the same name from the `math` module:

````python
import math
math.factorial(5)
````

````python
from math import cos, sin
from math import pi as PI
x = lambda r, phi, theta: r * cos(phi) * sin(theta)
x(1, 0, PI/2)
````
